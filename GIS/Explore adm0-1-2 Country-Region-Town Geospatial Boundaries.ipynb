# Geospatial Boundaries Interactive Coverage Map

**Goal**: Understand which areas of the globe are covered by ADM0/1/2 boundaries (and coloured), and which are not.
- ADM0 = country
- ADM1 = within-country region
- ADM2 = town

Input dataset in zip format:
* From https://github.com/wmgeolab/geoBoundaries/tree/main/releaseData/CGAZ
  * Dataset collected by William & Mary geoLab
  * The geoBoundaries Global Database of Political Administrative Boundaries Database is an online, open license (CC BY 4.0 / ODbL) resource of information on administrative boundaries for every country in the world

## ADM0 = country

```python
import geopandas as gpd
import requests
import io
import folium
import zipfile

# 1. Download ADM0 CGAZ directly into memory
def load_minimal_adm0():
    url = "https://github.com/wmgeolab/geoBoundaries/raw/main/releaseData/CGAZ/geoBoundariesCGAZ_ADM0.zip"
    print("Downloading boundary data...")
    r = requests.get(url)
    z = zipfile.ZipFile(io.BytesIO(r.content))
    # Read the shapefile component from the zip using fiona engine
    return gpd.read_file(io.BytesIO(r.content), engine='fiona')

gdf_adm0 = load_minimal_adm0()

# 2. Simplify and Plot
print("Generating interactive map...")
gdf_simple = gdf_adm0.copy()
gdf_simple['geometry'] = gdf_simple['geometry'].simplify(tolerance=0.02, preserve_topology=True)

m = folium.Map(location=[20, 0], zoom_start=2, tiles='CartoDB positron')

folium.GeoJson(
    gdf_simple,
    name="geoBoundaries ADM0",
    style_function=lambda x: {
        'fillColor': '#228B22',
        'color': '#006400',
        'weight': 1,
        'fillOpacity': 0.4
    },
    tooltip=folium.GeoJsonTooltip(fields=['shapeName', 'shapeGroup'], aliases=['Country:', 'ISO:'])
).add_to(m)

folium.LayerControl().add_to(m)
display(m)
```

## ADM1 = within-country region

```python
import geopandas as gpd
import requests
import io
import folium
import zipfile

# 1. Download ADM1 CGAZ directly into memory
def load_minimal_adm1():
    url = "https://github.com/wmgeolab/geoBoundaries/raw/main/releaseData/CGAZ/geoBoundariesCGAZ_ADM1.zip"
    print("Downloading ADM1 boundary data (Regions)...")
    r = requests.get(url)
    # Read directly using fiona engine
    return gpd.read_file(io.BytesIO(r.content), engine='fiona')

gdf_adm1 = load_minimal_adm1()

# 2. Simplify and Plot
print("Generating interactive ADM1 map...")
gdf_simple1 = gdf_adm1.copy()
gdf_simple1['geometry'] = gdf_simple1['geometry'].simplify(tolerance=0.02, preserve_topology=True)

m1 = folium.Map(location=[20, 0], zoom_start=2, tiles='CartoDB positron')

folium.GeoJson(
    gdf_simple1,
    name="geoBoundaries ADM1",
    style_function=lambda x: {
        'fillColor': '#4169E1',
        'color': '#00008B',
        'weight': 1,
        'fillOpacity': 0.4
    },
    tooltip=folium.GeoJsonTooltip(fields=['shapeName', 'shapeGroup'], aliases=['Region:', 'ISO:'])
).add_to(m1)

folium.LayerControl().add_to(m1)
display(m1)
```

## ADM2 = town  

```python
import geopandas as gpd
import requests
import io
import folium
import zipfile

# 1. Download ADM2 CGAZ directly into memory (Note: This file is >800MB extracted)
def load_minimal_adm2():
    url = "https://github.com/wmgeolab/geoBoundaries/raw/main/releaseData/CGAZ/geoBoundariesCGAZ_ADM2.zip"
    print("Downloading ADM2 boundary data (Towns) - this may take a moment...")
    r = requests.get(url)
    return gpd.read_file(io.BytesIO(r.content), engine='fiona')

gdf_adm2 = load_minimal_adm2()

# 2. Simplify significantly for performance
print("Generating interactive ADM2 map...")
gdf_simple2 = gdf_adm2.copy()
gdf_simple2['geometry'] = gdf_simple2['geometry'].simplify(tolerance=0.05, preserve_topology=True)

m2 = folium.Map(location=[20, 0], zoom_start=2, tiles='CartoDB positron')

folium.GeoJson(
    gdf_simple2,
    name="geoBoundaries ADM2",
    style_function=lambda x: {
        'fillColor': '#FF8C00',
        'color': '#8B4513',
        'weight': 0.5,
        'fillOpacity': 0.3
    },
    tooltip=folium.GeoJsonTooltip(fields=['shapeName', 'shapeGroup'], aliases=['Town:', 'ISO:'])
).add_to(m2)

folium.LayerControl().add_to(m2)
display(m2)
```
